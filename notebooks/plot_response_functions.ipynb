```python
def process_response_function(response_function, pad_to=81, modified_half_length=40) -> torch.Tensor:        
    padding_size = modified_half_length
    response_function = torch.nn.functional.interpolate(response_function, size=2*padding_size+1, mode='linear')
    response_function /= response_function.sum()
    if pad_to is not None:
        pad_size_left = (pad_to - response_function.shape[-1]) // 2
        pad_size_right = pad_to - response_function.shape[-1] - pad_size_left
        response_function = torch.nn.functional.pad(response_function, (pad_size_left, pad_size_right))
    return response_function
```

```python
idx = 0
original_response = process_response_function(original_responses[idx,0,0])[0,0].numpy()
smoothed_response = process_response_function(smoothed_responses[idx,0,0])[0,0].numpy()
altered_response = process_response_function(smoothed_responses[idx,0,0], modified_half_length=round(1.3*40))[0,0].numpy()

plt.figure(figsize=(10,6))
plt.plot(original_response, label='Original Response Function', linewidth=2)
plt.plot(smoothed_response, label='Smoothed Response Function', linewidth=2)
plt.plot(altered_response, label='Altered Response Function', linewidth=2)
plt.title('Comparison of Response Functions', fontsize=16)
```
