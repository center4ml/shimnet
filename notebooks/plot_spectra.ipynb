```python
import pandas as pd
import plotly.graph_objects as go
import numpy as np
from pathlib import Path
```

```python
def load_spectrum(path):
    df = pd.read_csv(path, sep=None, engine="python", comment="#", header=None)
    df = df.dropna(axis=1, how="all")  # usuń puste kolumny
    freqs = df.iloc[:, 0].values.astype(float)
    intensities = df.iloc[:, 1].values.astype(float)
    return freqs, intensities
```

```python
# spectrum_name = "mieszanina_poReakcji_mono-click"
spectrum_name = "mono-click_z1-80_z2+50"

# checkpoint_dir = "singlets_fixed_positions-nw_fixed_height_hotstart_long/plots/0_3908"
# checkpoint_dir = "singlets_fixed_positions-nw_fixed_height_hotstart_long/plots/0_5862"
checkpoint_dir = "singlets_fixed_positions-nw_fixed_height_hotstart_long/plots/0_29310"
reference_name = ""#"14Nov2025_pierwsze"


fig = go.Figure()

freqs_input, spec_input = load_spectrum(f"/scratch/mbuk/nmr/projects/01_shimator/mixture/spectra/input/{spectrum_name}.csv")
freqs_corrected, spec_corrected = load_spectrum(f"/data_hdd/mbuk/nmr/projects/01_shimator/shimnet/runs/from_peak_list/{checkpoint_dir}/extra_spectra/{spectrum_name}_clean.csv")
# freqs_reference, spec_reference = load_spectrum(f"/scratch/mbuk/nmr/projects/01_shimator/mixture/spectra/output/{spectrum_name}_{reference_name}.csv") #fixnw.csv")
# freqs_reference, spec_reference = load_spectrum("/scratch/mbuk/nmr/projects/01_shimator/mixture/spectra/input/14Nov2025_pierwsze.csv") #fixnw.csv")
freqs_reference, spec_reference = load_spectrum("/scratch/mbuk/nmr/projects/01_shimator/mixture/spectra/input/14Nov2025_poreakcji.csv") #fixnw.csv")


fig.add_trace(go.Scatter(
    x=freqs_input,
    y=spec_input / spec_input.max(),
    mode='lines',
    name="input",
    # line=dict(color='#EF553B')
))
fig.add_trace(go.Scatter(
    x=freqs_corrected,
    y=spec_corrected / spec_corrected.max(),
    mode='lines',
    name=f"corrected {checkpoint_dir}",
    # line=dict(color='#00CC96')
))
fig.add_trace(go.Scatter(
    x=freqs_reference,
    y=spec_reference / spec_reference.max(),
    mode='lines',
    name=f"reference {reference_name}",
    # line=dict(color='#636EFA')
))

fig.update_xaxes(autorange="reversed")

fig.update_layout(title=spectrum_name)
```

```python
blind_predictions = True
save_plots = True
output_dir_main = Path("../../mixture/output_plots")
```

```python
for input_spectrum_path in Path("../../mixture/spectra/input").glob("*.csv"):
    output_label = input_spectrum_path.stem + ("_blind" if blind_predictions else "")
    output_dir = output_dir_main / output_label
    predictions_paths = [p.resolve() for p in Path("../../mixture/spectra/output").glob(f"{input_spectrum_path.stem}*csv")]

    if save_plots:
        output_dir.mkdir(exist_ok=True, parents=True)
    
    if blind_predictions:
        precition_lables = np.arange(1, 1+len(predictions_paths))
        np.random.shuffle(precition_lables)

        if save_plots:
            with open(output_dir / f"{output_label}_labels.txt", "w") as f:
                for i, path in enumerate(predictions_paths):
                    f.write(f"{precition_lables[i]} {path.name}\n")
    
    fig = go.Figure()

    freqs_input, spec_input = load_spectrum(input_spectrum_path)
    fig.add_trace(go.Scatter(
        x=freqs_input,
        y=spec_input,
        mode='lines',
        name="input",
        # line=dict(color='#EF553B')
    ))

    for i_spectrum, spectrum_path in enumerate(predictions_paths):
        freqs_input, spec_input = load_spectrum(spectrum_path)

        
        fig.add_trace(go.Scatter(
            x=freqs_input,
            y=spec_input,
            mode='lines',
            name=f"prediction {precition_lables[i_spectrum]}" if blind_predictions else spectrum_path.stem,
            # line=dict(color='#EF553B')
        ))

    # odwrócenie osi X
    fig.update_xaxes(autorange="reversed")

    if save_plots:
        fig.write_html(output_dir / f"{output_label}_spectra.html")

```
