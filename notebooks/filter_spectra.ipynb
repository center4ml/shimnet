```python
input_dir = Path("../../mixture/spectra/input_full")
output_dir = Path("../../mixture/spectra/input")
output_dir.mkdir(parents=True, exist_ok=True)

min_freq = 4
max_freq = 8

for f in input_dir.glob("*.csv"):
    data = np.loadtxt(f)
    freqs = data[:, 0]
    spec = data[:, 1]

    filter_mask = (freqs >= min_freq) & (freqs <= max_freq)
    freqs_filtered = freqs[filter_mask]
    spec_filtered = spec[filter_mask]

    np.savetxt(output_dir / f.name, np.column_stack((freqs_filtered, spec_filtered)))
    
    # # Simple low-pass filter: zero out frequencies above a threshold
    # threshold = 5000  # Adjust this value as needed
    # spec_filtered = np.where(np.abs(freqs) > threshold, 0, spec)
    
    # output_data = np.column_stack((freqs, spec_filtered))
    # np.savetxt(output_dir / f.name, output_data, delimiter=",")
```
