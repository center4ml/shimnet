```python
import numpy as np
import matplotlib.pyplot as plt
import torch
from pathlib import Path
```

```python
run_dir = Path("../runs/monitorowanie_reakcji/29_reakcja2_dotrenowanie_wszystkieZ2")

logs = np.loadtxt(run_dir / "train.txt")
columns = ["BatchInStage", "Loss", "AvgLoss", "CleanLoss", "RespLoss", "NoisedLoss"]
# len(losses)

selected_column = 2
selected_column_name = columns[selected_column]
skip_first = 100 #130000

fig, ax = plt.subplots()
plt.title(f'Training Progress: {selected_column_name}', fontsize=16)
ax.scatter(np.arange(len(logs[skip_first:,selected_column])), logs[skip_first:,selected_column], marker=".", alpha=0.1)
plt.semilogy(np.minimum.accumulate(logs[skip_first:,selected_column], axis=0), c="C0", alpha=0.5, label=f'Best Training {selected_column_name}')
plt.semilogy(np.maximum.accumulate(logs[skip_first:,selected_column][::-1])[::-1], c="C0", alpha=0.5, label=f'Worst Future Training {selected_column_name}')
```
